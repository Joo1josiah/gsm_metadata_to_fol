```python
gse = GEOparse.get_GEO("GSE1567", destdir="./", silent=True)
```

```python
gsms = gse.gsms
print(f"Number of samples: {len(gsms)}")

gsm_metadata = {gsm: gsms[gsm].metadata for gsm in gsms}
first_gsm = next(iter(gsms))  # Get the first key from the gsms dictionary
print(f"Metadata for first sample: {gsm_metadata[first_gsm]}")
```

```python
def metadata_to_fol(metadata):
    fol_statements = []
    sample_id = metadata.get("geo_accession", ["Unknown"])[0]  # Get the sample ID

    for key, value in metadata.items():

        predicate = key.replace(" ", "_").replace("/", "_").replace("-", "_")
        value_str = ", ".join(f'"{v}"' if isinstance(v, str) else str(v) for v in value)
        fol_statements.append(f"({sample_id} {predicate} {value_str})")
    return fol_statements

fol_statements = metadata_to_fol(gsm_metadata)

# Print the FOL statements
for statement in fol_statements:
    print(statement)
```

```python
def metadata_to_fol_all(gsm_metadata):

    all_fol_statements = {}

    for sample_id, metadata in gsm_metadata.items():
        fol_statements = []
        for key, value in metadata.items():
            predicate = key.replace(" ", "_").replace("/", "_").replace("-", "_")
            value_str = ", ".join(f'"{v}"' if isinstance(v, str) else str(v) for v in value)
            fol_statements.append(f"({sample_id} {predicate} {value_str})")
        # Store the FOL statements for the current sample
        all_fol_statements[sample_id] = fol_statements

    return all_fol_statements

# Generate FOL statements for all samples
all_fol_statements = metadata_to_fol_all(gsm_metadata)

# Print FOL statements for each sample
for sample_id, fol_statements in all_fol_statements.items():
    print(f"FOL statements for {sample_id}:")
    for statement in fol_statements:
        print(statement)
    print("\n")
```
